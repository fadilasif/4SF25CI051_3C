## Data cleanup

```python
df = df.drop(columns=[
    'url',
    'address',
    'book_table',
    'phone',
    'reviews_list',
    'menu_item',
    'listed_in(type)',
    'listed_in(city)'
])
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python

df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
df = df.drop(columns=['dish_liked'])
```

```python
df.isnull().sum()
```

```python
print("Total Missing Values:", df.isnull().sum().sum())
```

```python
df = df.reset_index(drop=True)
```

## Handle NULL values


```python
df.head()
```

```python
df.tail()
```

```python
df['rate'] = df['rate'].astype(str).str.replace('/5', '')
df['rate'] = df['rate'].replace(['NEW', '-'], np.nan)

df['rate'] = pd.to_numeric(df['rate'], errors='coerce')
```

```python
df['approx_cost(for two people)'] = df['approx_cost(for two people)'].astype(str).str.replace(',', '')
df['approx_cost(for two people)'] = pd.to_numeric(df['approx_cost(for two people)'], errors='coerce')
```

```python
# Numeric values with mean
df['rate'] = df['rate'].fillna(df['rate'].mean())

df['approx_cost(for two people)'] = df['approx_cost(for two people)'].fillna(df['approx_cost(for two people)'].mean())
```

```python
# Categories with mode
df['location'] = df['location'].fillna(df['location'].mode()[0])
df['rest_type'] = df['rest_type'].fillna(df['rest_type'].mode()[0])
df['cuisines'] = df['cuisines'].fillna(df['cuisines'].mode()[0])
```

```python
print(df.isnull().sum())
print("Total Missing Values:", df.isnull().sum().sum())
```

```python
# Calculate cost per person
df['cost_per_person'] = df['approx_cost(for two people)'] / 2
```

## Data visualization

```python
# top 10 restuarant locations
top_locations = df['location'].value_counts().head(10)
plt.figure(figsize=(10, 5))
top_locations.plot(kind='bar')
plt.title('Top 10 Restaurant Locations')
plt.xlabel('Location')
plt.ylabel('Number of Restaurants')
plt.xticks(rotation=45)
plt.show()
```

```python
# top 10 cuisines
top_cuisines = df['cuisines'].value_counts().head(10)

plt.figure(figsize=(10, 5))
top_cuisines.plot(kind='bar')

plt.title('Top 10 Popular Cuisines')
plt.xlabel('Cuisine')
plt.ylabel('Number of Restaurants')
plt.xticks(rotation=45)
plt.show()
```

```python
# Online vs offline ordering
plt.figure(figsize=(6, 6))
df['online_order'].value_counts().plot(kind='pie', autopct='%1.1f%%')

plt.title('Online vs Offline Ordering')
plt.ylabel('')
plt.show()
```

```python
# Restaurant rating distribution
plt.figure(figsize=(8, 5))
plt.hist(df['rate'], bins=10)

plt.title('Restaurant Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Number of Restaurants')
plt.show()
```

```python
# Restaurant cost distribution
plt.figure(figsize=(8, 5))
plt.hist(df['approx_cost(for two people)'], bins=20)

plt.title('Restaurant Cost Distribution')
plt.xlabel('Cost for Two People')
plt.ylabel('Number of Restaurants')
plt.show()
```

```python
# top 10 restaurants by vote
restaurant_votes = df.groupby('name')['votes'].sum().reset_index()

restaurant_votes = restaurant_votes.sort_values(
    by='votes', ascending=False
).head(10)

plt.figure(figsize=(10, 6))
sns.barplot(data=restaurant_votes, x='votes', y='name')

plt.title('Top 10 Restaurants by Total Votes')
plt.xlabel('Total Votes')
plt.ylabel('Restaurant')
plt.show()
```
